## 3. Lọc kết quả

### 3.1. Lọc kết quả sử dụng find()

```python
# Đối số đầu tiên của phương thức find() là một đối tượng truy vấn và được
# sử dụng để giới hạn tìm kiếm.
my_query = {"age": 22}
my_docs = my_col.find(my_query)

for x in my_docs:
    print(x)
```

### 3.2. Lọc kết quả thay đổi giá trị đối tượng truy vấn

```python
# Sử dụng "$gt" để lấy giá trị lớn hơn trong đối tượng truy vấn
my_query = {"age": {"$gt": 25}}
fields = {"_id": 0}

my_docs = my_col.find(my_query, fields)

for x in my_docs:
    print(x)
```

**Các toán tử truy vấn so sánh trong MongoDB**

| Tên    | Mô tả                                                             |
| ------ | ----------------------------------------------------------------- |
| `$eq`  | Khớp với các giá trị bằng một giá trị được chỉ định.              |
| `$gt`  | Khớp với các giá trị lớn hơn một giá trị được chỉ định.           |
| `$gte` | Khớp với các giá trị lớn hơn hoặc bằng một giá trị được chỉ định. |
| `$in`  | Khớp với bất kỳ giá trị nào được chỉ định trong một mảng.         |
| `$lt`  | Khớp với các giá trị nhỏ hơn một giá trị được chỉ định.           |
| `$lte` | Khớp với các giá trị nhỏ hơn hoặc bằng một giá trị được chỉ định. |
| `$ne`  | Khớp với tất cả các giá trị không bằng một giá trị được chỉ định. |
| `$nin` | Không khớp với giá trị nào được chỉ định trong mảng.              |


```python
# Sử dụng "$gt" để lấy giá trị lớn hơn trong đối tượng truy vấn
my_query = {"age": {"$gt": 20}}
fields = {"_id": 0}

my_docs = my_col.find(my_query, fields)

for x in my_docs:
    print(x)
```

```python
# $in: Tìm giá trị nằm trong một danh sách (List).
my_query = {"age": {"$in": [22, 25, 30]}}
fields = {"_id": 0}

my_docs = my_col.find(my_query, fields)

for x in my_docs:
    print(x)
```

```python
# Tìm kiếm chuỗi tương đối (Tương đương LIKE trong SQL)
my_query = {"name": {"$regex": "An", "$options":"i"}}
fields = {"_id": 0}

my_docs = my_col.find(my_query, fields)

for x in my_docs:
    print(x)
```

**Các toán tử logic so sánh trong MongoDB**

| Tên    | Mô tả                                                                                                                 |
| ------ | --------------------------------------------------------------------------------------------------------------------- |
| `$and` | Nối các mệnh đề truy vấn với AND logic trả về tất cả các tài liệu phù hợp với điều kiện của cả hai mệnh đề.           |
| `$not` | Đảo ngược mệnh đề truy vấn và trả về các tài liệu không khớp với mệnh đề truy vấn.                                    |
| `$nor` | Nối các mệnh đề truy vấn với NOR logic trả về tất cả các tài liệu không khớp với cả hai mệnh đề.                      |
| `$or`  | Nối các mệnh đề truy vấn với OR logic trả về tất cả các tài liệu phù hợp với các điều kiện của một trong hai mệnh đề. |


```python
# Logic AND: Tìm sinh viên nam có tuổi lớn hơn 20
my_query = {"gender": "Nam", "age":{"$gt":20}}
fields = {"_id": 0}

my_docs = my_col.find(my_query, fields)

for x in my_docs:
    print(x)
```

```python
# Kết hợp AND và OR:
# Tìm sinh viên nam có tuổi bằng 25 hoặc 30
my_query = {"gender": "Nam", "$or":[{"age":25}, {"age":30}]}
fields = {"_id": 0}

my_docs = my_col.find(my_query, fields)

for x in my_docs:
    print(x)
```

## 4. Sắp xếp kết quả

Phương thức sort() lấy một tham số cho tên trường và một tham số là chiều sắp xếp (mặc định là tăng dần).

```python
my_docs = my_col.find({},
    {"_id": 0,}).sort("age", 1)

for x in my_docs:
    print(x)
```

## 5. Giới hạn kết quả

Để giới hạn kết quả trong MongoDB, chúng ta sử dụng phương thức limit().

```python
# Phương thức limit() có một tham số xác định số lượng tài liệu sẽ trả về.
my_docs = my_col.find({}, {"_id":0}).limit(2)

for x in my_docs:
    print(x)
```

```python
# Sắp xếp và giới hạn
my_docs = my_col.find({},{"_id": 0,}).sort("age", -1).limit(5)
for x in my_docs:
    print(x)
```

```python
count = my_col.count_documents({"age": 30})

print(f"Số sinh viên có tuổi bằng 30: {count}")
```

## 6. Đối chiếu SQL với MongoDB

| Yêu cầu        | SQL                                 | MongoDB (PyMongo)                             |
| -------------- | ----------------------------------- | --------------------------------------------- |
| Lấy tất cả     | `SELECT * FROM students`            | `find({})`                                    |
| Lọc cơ bản     | `WHERE age = 20`                    | `find({"age": 20})`                           |
| Lọc AND        | `WHERE age > 20 AND gender = 'Nam'` | `find({"age": {"$gt": 20}, "gender": "Nam"})` |
| Tìm chứa chuỗi | `WHERE name LIKE '%An%'`            | `find({"name": {"$regex": "An"}})`            |
| Đếm số lượng   | `SELECT COUNT(*) FROM students`     | `count_documents({})`                         |
| Sắp xếp        | `ORDER BY age DESC`                 | `find().sort("age", -1)`                      |
| Giới hạn       | `LIMIT 5`                           | `find().limit(5)`                             |


## 7. Tổng kết

```python
# Level 1 – Dễ: find_one() theo student_id
student = my_col.find_one({"student_id": "SV001"})
if student is None:
    print("Không tìm thấy sinh viên nào khớp điều kiện")
else:
    print(student)  # Hiển thị Dictionary nguyên bản
```

```python
# Level 2 – Trung bình: AND + sort() + limit(2) + f-string
query = {"major": "IT", "gpa": {"$gt": 3.0}}
cursor = my_col.find(query).sort("gpa", -1).limit(2)
found = False
for student in cursor:
    found = True
    print(f"Sinh viên: {student['name']} - Điểm GPA: {student['gpa']}")
if not found:
    print("Không tìm thấy sinh viên nào khớp điều kiện")
```

```python
# Level 3 – Khó: $or + $in + Projection
query = {
    "$or": [{"major": "IT"}, {"major": "Data"}],
    "skills": {"$in": ["Python", "Java"]},
}
projection = {"_id": 0, "name": 1, "skills": 1}
cursor = my_col.find(query, projection)
found = False
for student in cursor:
    found = True
    print(f"Sinh viên: {student['name']} - Kỹ năng: {', '.join(student['skills'])}")
if not found:
    print("Không tìm thấy sinh viên nào khớp điều kiện")
```
